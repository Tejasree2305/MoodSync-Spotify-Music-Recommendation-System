### 3.51 Creating a Reusable Mood-Based Recommendation Function

In this step, we combine mood filtering,
popularity sorting, and duplicate removal
into a reusable Python function.

```python

# Create a reusable mood-based recommendation function
def recommend_by_mood(mood, top_n=10):

    # Filter songs by mood
    songs = df_unique[
        df_unique['mood'].str.lower() == mood.strip().lower()
    ]

    # Sort songs by popularity
    songs = songs.sort_values(
        by='popularity',
        ascending=False
    )

    # Remove repeated song and artist combinations
    songs = songs.drop_duplicates(
        subset=['track_name', 'artists'],
        keep='first'
    )

    # Return the top matching songs
    return songs[
        ['track_name', 'artists', 'popularity', 'mood']
    ].head(top_n)


# Test the function
recommend_by_mood('Happy / Energetic', 10)
```

### 3.55 Calculating Cosine Similarity Between Two Songs

In this step, we use NumPy to calculate
the cosine similarity between the first
two songs in our Spotify dataset.

```python

# Select the first two song vectors
song1 = feature_vectors[0]
song2 = feature_vectors[1]

# Calculate the dot product
dot_product = np.dot(song1, song2)

# Calculate the length of each vector
magnitude1 = np.linalg.norm(song1)
magnitude2 = np.linalg.norm(song2)

# Calculate cosine similarity
if magnitude1 == 0 or magnitude2 == 0:
    print("Cosine similarity is undefined for a zero vector.")
else:
    cosine_similarity = dot_product / (magnitude1 * magnitude2)

    print("Song 1:", df_unique.iloc[0]['track_name'])
    print("Song 2:", df_unique.iloc[1]['track_name'])
    print("Cosine Similarity:", round(cosine_similarity, 4))
```

### 3.56 Finding Similar Songs Using Cosine Similarity

In this step, we calculate cosine similarity
between one selected song and all songs
in our dataset using NumPy.

```python

# Select the first song as our reference
selected_song = feature_vectors[0]

# Calculate dot products with all songs
dot_products = np.dot(feature_vectors, selected_song)

# Calculate the magnitude of every song vector
all_magnitudes = np.linalg.norm(feature_vectors, axis=1)

# Calculate the magnitude of the selected song
selected_magnitude = np.linalg.norm(selected_song)

# Calculate cosine similarity safely
denominators = all_magnitudes * selected_magnitude

similarity_scores = np.divide(
    dot_products,
    denominators,
    out=np.zeros_like(dot_products),
    where=denominators != 0
)

# Display the results
print("Selected Song:", df_unique.iloc[0]['track_name'])
print("Total Similarity Scores:", len(similarity_scores))
print("First 5 Similarity Scores:")
print(np.round(similarity_scores[:5], 4))
```

### 3.57 Finding the Top 10 Most Similar Songs

In this step, we rank songs based on their

```python

# Sort similarity scores from highest to lowest
sorted_indices = np.argsort(similarity_scores)[::-1]

# Remove the selected song itself
sorted_indices = sorted_indices[sorted_indices != 0]

# Select the top 10 similar songs
top_indices = sorted_indices[:10]

# Get the recommended songs
similar_songs = df_unique.iloc[top_indices].copy()

# Add cosine similarity scores
similar_songs['similarity_score'] = similarity_scores[top_indices]

# Display the recommendations
print("Recommendations for:", df_unique.iloc[0]['track_name'])

display(similar_songs[
    ['track_name', 'artists', 'mood', 'similarity_score']
])
```

### 3.59 Creating a Reusable Content-Based Recommendation Function

In this step, we create a Python function that
accepts a song title, calculates cosine similarity,
and returns the most similar songs.

```python
### 3.59 Creating a Reusable Content-Based Recommendation Function

In this step, we create a Python function that
accepts a song title, calculates cosine similarity,
and returns the most similar songs.
```

```python

# Create a reusable content-based recommendation function
def recommend_similar_songs(song_name, top_n=10):

    # Find songs matching the entered title
    matches = df_unique[
        df_unique['track_name'].str.lower() == song_name.strip().lower()
    ]

    # Check whether the song exists
    if matches.empty:
        print("Song not found. Please try another title.")
        return

    # Get the position of the first matching song
    song_index = df_unique.index.get_indexer([matches.index[0]])[0]

    # Select the song's audio feature vector
    selected_vector = feature_vectors[song_index]

    # Calculate cosine similarity with all songs
    dot_products = np.dot(feature_vectors, selected_vector)

    all_magnitudes = np.linalg.norm(feature_vectors, axis=1)
    selected_magnitude = np.linalg.norm(selected_vector)

    denominators = all_magnitudes * selected_magnitude

    scores = np.divide(
        dot_products,
        denominators,
        out=np.zeros_like(dot_products),
        where=denominators != 0
    )

    # Sort songs from highest to lowest similarity
    sorted_positions = np.argsort(scores)[::-1]

    # Exclude the selected song
    sorted_positions = sorted_positions[sorted_positions != song_index]

    # Get ranked recommendations
    recommendations = df_unique.iloc[sorted_positions].copy()

    # Add similarity scores
    recommendations['similarity_score'] = scores[sorted_positions]

    # Remove repeated song-and-artist combinations
    recommendations = recommendations.drop_duplicates(
        subset=['track_name', 'artists'],
        keep='first'
    )

    # Exclude the selected title and artist
    selected_title = df_unique.iloc[song_index]['track_name']
    selected_artist = df_unique.iloc[song_index]['artists']

    recommendations = recommendations[
        ~(
            (recommendations['track_name'] == selected_title) &
            (recommendations['artists'] == selected_artist)
        )
    ]

    # Return the top recommendations
    return recommendations[
        ['track_name', 'artists', 'mood', 'similarity_score']
    ].head(top_n)


# Test the function
recommend_similar_songs('Comedy', 10)
```

### 3.60 Testing Content-Based Recommendations with Different Songs

### 3.70 Measuring Audio-Feature Diversity

In this step, we calculate the average pairwise
Euclidean distance among recommended songs.

We use NumPy to measure how different the
recommended songs are from one another.

```python

# Get the seven audio features of recommended songs
recommended_vectors = df_unique.loc[
    results.index, audio_features
].to_numpy(dtype=float)

# Store distances between different song pairs
pairwise_distances = []

# Compare each song with every later song
for i in range(len(recommended_vectors)):
    for j in range(i + 1, len(recommended_vectors)):

        # Calculate Euclidean distance
        distance = np.linalg.norm(
            recommended_vectors[i] - recommended_vectors[j]
        )

        pairwise_distances.append(distance)

# Calculate average audio-feature diversity
if len(pairwise_distances) > 0:
    audio_diversity = np.mean(pairwise_distances)
else:
    audio_diversity = 0

# Display results
print("Audio-Feature Diversity Evaluation")
print("----------------------------------")
print("Selected Song:", user_song)
print("Recommended Songs:", len(recommended_vectors))
print("Number of Song Pairs:", len(pairwise_distances))
print("Average Pairwise Distance:", round(audio_diversity, 4))
```

### 3.71 Comparing Cosine Similarity and Euclidean Distance

In this step, we compare cosine similarity and
Euclidean distance for the same five recommended songs.

We use NumPy and Pandas to examine how the two
measures describe audio-feature relationships.

```python

# Select Comedy's audio feature vector
comedy_vector = feature_vectors[0]

# Get the five recommended songs' feature vectors
recommendation_vectors = df_unique.loc[
    results.index, audio_features
].to_numpy(dtype=float)

# Calculate Euclidean distance from Comedy
euclidean_distances = np.linalg.norm(
    recommendation_vectors - comedy_vector,
    axis=1
)

# Create a comparison table
metric_comparison = pd.DataFrame({
    'Song': results['track_name'].values,
    'Cosine Similarity': results['similarity_score'].values,
    'Euclidean Distance': euclidean_distances
})

# Display the comparison
print("Cosine Similarity vs Euclidean Distance")
display(metric_comparison.round(4))
```

### 3.72 Comparing Recommendation Rankings

In this step, we compare the rankings produced

```python

# Copy the existing comparison table
ranking_comparison = metric_comparison.copy()

# Rank songs by cosine similarity (highest first)
ranking_comparison['Cosine Rank'] = (
    ranking_comparison['Cosine Similarity']
    .rank(ascending=False, method='min')
    .astype(int)
)

# Rank songs by Euclidean distance (lowest first)
ranking_comparison['Euclidean Rank'] = (
    ranking_comparison['Euclidean Distance']
    .rank(ascending=True, method='min')
    .astype(int)
)

# Sort by cosine similarity rank
ranking_comparison = ranking_comparison.sort_values(
    by='Cosine Rank'
)

# Display the rankings
print("Recommendation Ranking Comparison:")
display(ranking_comparison.round(4))
```

### 3.77 Creating a Reusable Evaluation Function

In this step, we create a Python function to
evaluate recommendations for different songs.

The function calculates:
- Total recommendations
- Average cosine similarity
- Number of unique mood categories

This makes our evaluation code reusable.

```python

# Create a reusable evaluation function
def evaluate_recommendations(song_name, top_n=5):

    # Generate recommendations
    recommendations = recommend_similar_songs(
        song_name,
        top_n=top_n
    )

    # Handle missing songs or empty recommendations
    if recommendations is None or recommendations.empty:
        print("No recommendations available.")
        return None

    # Calculate evaluation metrics
    total_songs = len(recommendations)

    avg_similarity = recommendations[
        'similarity_score'
    ].mean()

    unique_mood_count = recommendations[
        'mood'
    ].nunique()

    # Display evaluation results
    print("Reusable Recommendation Evaluation")
    print("----------------------------------")
    print("Selected Song:", song_name)
    print("Total Recommendations:", total_songs)
    print("Average Cosine Similarity:", round(avg_similarity, 4))
    print("Unique Mood Categories:", unique_mood_count)

    return recommendations


# Test the function using Comedy
test_results = evaluate_recommendations(
    "Comedy",
    top_n=5
)
```

### 3.83 Building an Automated Test Summary

In this step, we create a structured test report
for our MoodSync recommendation evaluation function.

We will:
- Test valid and invalid inputs
- Record expected and actual behavior
- Calculate the total number of passed tests
- Display the results using Pandas

```python

# Define automated test cases
test_cases = [
    ("Empty Song", "", 5, False),
    ("Spaces Only", "   ", 5, False),
    ("Numeric Title", 12345, 5, False),
    ("Missing Title", None, 5, False),
    ("Zero Count", "Comedy", 0, False),
    ("Negative Count", "Comedy", -3, False),
    ("Decimal Count", "Comedy", 2.5, False),
    ("Valid Song", "Comedy", 5, True)
]

# Store test results
test_records = []

for test_name, song, count, expected_valid in test_cases:

    print("\nRunning Test:", test_name)

    try:
        result = evaluate_recommendations(
            song_name=song,
            top_n=count
        )

        actual_valid = result is not None

        passed = actual_valid == expected_valid

        test_records.append({
            "Test Name": test_name,
            "Expected Valid": expected_valid,
            "Actual Valid": actual_valid,
            "Status": "PASSED" if passed else "FAILED"
        })

    except Exception as error:
        test_records.append({
            "Test Name": test_name,
            "Expected Valid": expected_valid,
            "Actual Valid": "ERROR",
            "Status": "FAILED"
        })

        print("Error:", str(error))


# Convert results into a DataFrame
test_summary = pd.DataFrame(test_records)

# Calculate test statistics
total_tests = len(test_summary)

passed_tests = (
    test_summary["Status"] == "PASSED"
).sum()

failed_tests = total_tests - passed_tests

pass_rate = (passed_tests / total_tests) * 100

# Display the final report
print("\nMoodSync - Automated Test Summary")
print("--------------------------------")
display(test_summary)

print("Total Tests:", total_tests)
print("Passed Tests:", passed_tests)
print("Failed Tests:", failed_tests)
print("Pass Rate:", round(pass_rate, 2), "%")
```
